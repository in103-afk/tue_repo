## Metacharacters that match a single character

```python
# [] represent character class, character class metacharacter sequence, will match any single character contained in the class 
#character class basically lets the target character match to any one of that group of characters which is in the box 
re.search('ba[artz]', 'foobarqux')
re.search('ba[artz]', 'foobazqux')
re.search('[a-z]', 'FOObar')
re.search('[0-9][0-9]', 'foo123bar')
#this class is given three possible ranges, for the thing to fall into 
re.search('[0-9a-fA-F]', '--- a0 ---')
#You can complement a character class by specifying ^ as the first character, in which case it matches any character that isn’t in the set. 
# in the following example, [^0-9] matches any character that isn’t a digit: 
re.search('[^0-9]', '12345foo')
#If a ^ character appears in a character class but isn’t the first character, 
# then it has no special meaning and matches a literal '^' character:
re.search('[#:^]', 'foo^bar:baz#qux')
#As you’ve seen, you can specify a range of characters in a character class by separating characters with a hyphen. What if you want the character class to include a literal hyphen character? 
# You can place it as the first or last character or escape it with a backslash (\):
re.search('[-abc]', '123-456')
re.search('[abc-]', '123-456')
re.search('[ab\-c]', '123-456')
#If you want to include a literal ']' in a character class, then you can place it as the first character or escape it with backslash:
re.search('[]]', 'foo[1]')
re.search('[ab\]cd]', 'foo[1]') 
#Other regex metacharacters lose their special meaning inside a character class:
re.search('[)*+|]', '123*456')
re.search('[)*+|]', '123+456')
#As you saw in the table above, * and + have special meanings in a regex in Python. 
# They designate repetition, which you’ll learn more about shortly. But in this example, they’re inside a character class, so they match themselves literally.
#The . metacharacter matches any single character except a newline:
re.search('foo.bar', 'fooxbar')
print(re.search('foo.bar', 'foobar'))
print(re.search('foo.bar', 'foo\nbar'))
#\w matches any alphanumeric word character. Word characters are uppercase and lowercase letters, digits, and the underscore (_) character, so \w is essentially shorthand for [a-zA-Z0-9_]:
re.search('\w', '#(.a$@&')
re.search('[a-zA-Z0-9_]', '#(.a$@&')
#In this case, the first word character in the string '#(.a$@&' is 'a'. (talking about above)

#\W is the opposite. It matches any non-word character and is equivalent to [^a-zA-Z0-9_]:
re.search('\W', 'a_1*3Qb')
re.search('[^a-zA-Z0-9_]', 'a_1*3Qb')
#Here, the first non-word character in 'a_1*3!b' is '*'.

#\d matches any decimal digit character. \D is the opposite. It matches any character that isn’t a decimal digit:
re.search('\d', 'abc4def')
re.search('\D', '234Q678')
#\d is essentially equivalent to [0-9], and \D is equivalent to [^0-9]

#\s matches any whitespace character:
re.search('\s', 'foo\nbar baz')

#\S is the opposite of \s. It matches any character that isn’t whitespace:
re.search('\S', '  \n foo  \n  ')

#Again, \s and \S consider a newline to be whitespace. In the example above, the first non-whitespace character is 'f'.
#The character class sequences \w, \W, \d, \D, \s, and \S can appear inside a square bracket character class as well:
re.search('[\d\w\s]', '---3---')
re.search('[\d\w\s]', '---a---')
re.search('[\d\w\s]', '--- ---')
#In this case, [\d\w\s] matches any digit, word, or whitespace character. 
# And since \w includes \d, the same character class could also be expressed slightly shorter as [\w\s].
```

```python
#When it’s not serving either of these purposes, the backslash escapes metacharacters. 
# A metacharacter preceded by a backslash loses its special meaning and matches the literal character instead. Consider the following examples: 
re.search('.', 'foo.bar')
re.search('\.', 'foo.bar')
#The . character in the <regex> on line 4 is escaped by a backslash, so it isn’t a wildcard. 
# It’s interpreted literally and matches the '.' at index 3 of the search string.
s = r'foo\bar'
print(s)

#two backlash gives an error 
#re.search('\\', s)
#two ways to get around this 
re.search('\\\\', s)
re.search(r'\\', s)
```

## Anchors 
### Anchors are zero-width matches. They don’t match any actual characters in the search string, and they don’t consume any of the search string during parsing. Instead, an anchor dictates a particular location in the search string where a match must occur.

^

```python
#In other words, regex ^foo stipulates that 'foo' must be present not just any old place in the search string, but at the beginning:
re.search('^foo', 'foobar')

#\A functions similarly:
re.search('\Afoo', 'foobar')
print(re.search('\Afoo', 'barfoo'))

#When the regex parser encounters $ or \Z, the parser’s current position must be at the end of the search string for it to find a match.
#  Whatever precedes $ or \Z must constitute the end of the search string:
re.search('bar$', 'foobar')
re.search('bar\Z', 'foobar')
re.search('bar$', 'foobar\n')
```

## Quantifiers 

```python
# A quantifier metacharacter immediately follows a portion of a <regex> and
# indicates how many times that portion must occur for the match to succeed.

# * matches zero or more repetitions of the preceding regex.
# For example, a* matches zero or more 'a' characters. That means it would match
# an empty string, 'a', 'aa', 'aaa', and so on.
re.search('foo-*bar', 'foobar')
# Zero dashes: <re.Match object; span=(0, 6), match='foobar'>
re.search('foo-*bar', 'foo-bar')
# One dash: <re.Match object; span=(0, 7), match='foo-bar'>
re.search('foo-*bar', 'foo--bar')
# Two dashes: <re.Match object; span=(0, 8), match='foo--bar'>
# The metacharacter sequence -* matches in all three cases.

# .* matches zero or more occurrences of any character. In other words, it
# essentially matches any character sequence up to a line break.
# (Remember that the . wildcard metacharacter doesn't match a newline.)
re.search('foo.*bar', '# foo $qux@grault % bar #')
# <re.Match object; span=(2, 23), match='foo $qux@grault % bar'>
# Once you start using quantifiers like *, the number of characters matched can be
# quite variable, and the information in the match object becomes more useful.

# + matches one or more repetitions of the preceding regex.
# This is similar to *, but the quantified regex must occur at least once.
print(re.search('foo-+bar', 'foobar'))
# Zero dashes: None
re.search('foo-+bar', 'foo-bar')
# One dash: <re.Match object; span=(0, 7), match='foo-bar'>
re.search('foo-+bar', 'foo--bar')
# Two dashes: <re.Match object; span=(0, 8), match='foo--bar'>
# The + metacharacter requires at least one occurrence of '-', so there isn't
# a match on the first one.

# ? matches zero or one repetitions of the preceding regex.
# There's only a match if the preceding regex occurs once or not at all.
re.search('foo-?bar', 'foobar')
# Zero dashes: <re.Match object; span=(0, 6), match='foobar'>
re.search('foo-?bar', 'foo-bar')
# One dash: <re.Match object; span=(0, 7), match='foo-bar'>
print(re.search('foo-?bar', 'foo--bar'))
# Two dashes: None

# Here are some more examples showing the use of all three quantifier metacharacters.
# This time, the quantified regex is the character class [1-9] instead of the
# simple character '-'.
re.match('foo[1-9]*bar', 'foobar')
# <re.Match object; span=(0, 6), match='foobar'>
re.match('foo[1-9]*bar', 'foo42bar')
# <re.Match object; span=(0, 8), match='foo42bar'>

print(re.match('foo[1-9]+bar', 'foobar'))
# None
re.match('foo[1-9]+bar', 'foo42bar')
# <re.Match object; span=(0, 8), match='foo42bar'>

re.match('foo[1-9]?bar', 'foobar')
# <re.Match object; span=(0, 6), match='foobar'>
print(re.match('foo[1-9]?bar', 'foo42bar'))
# None

# *?, +?, ?? are the non-greedy (or lazy) versions of the *, +, and ? quantifiers.
# When used alone, the quantifier metacharacters *, +, and ? are all greedy,
# meaning they produce the longest possible match.
re.search('<.*>', '%<foo> <bar> <baz>%')
# <re.Match object; span=(1, 18), match='<foo> <bar> <baz>'>
# The regex <.*> effectively means: a '<' character, then any sequence of
# characters, then a '>' character. Since the * metacharacter is greedy, it
# dictates the longest possible match, which includes everything up to and
# including the '>' character that follows 'baz'.

# If you want the shortest possible match instead, then use the non-greedy
# metacharacter sequence *?
re.search('<.*?>', '%<foo> <bar> <baz>%')
# <re.Match object; span=(1, 6), match='<foo>'>
# In this case, the match ends with the '>' character following 'foo'.

# Note: You could accomplish the same thing with the regex <[^>]*>, which means:
# a '<' character, then any sequence of characters other than '>', then a '>' character.
re.search('<[^>]*>', '%<foo> <bar> <baz>%')
# <re.Match object; span=(1, 6), match='<foo>'>

# There are lazy versions of the + and ? quantifiers as well.
re.search('<.+>', '%<foo> <bar> <baz>%')
# <re.Match object; span=(1, 18), match='<foo> <bar> <baz>'>
re.search('<.+?>', '%<foo> <bar> <baz>%')
# <re.Match object; span=(1, 6), match='<foo>'>

re.search('ba?', 'baaaa')
# <re.Match object; span=(0, 2), match='ba'>
re.search('ba??', 'baaaa')
# <re.Match object; span=(0, 1), match='b'>
# The greedy version, ?, matches one occurrence, so ba? matches 'b' followed by
# a single 'a'. The non-greedy version, ??, matches zero occurrences, so ba??
# matches just 'b'.

# {m} matches exactly m repetitions of the preceding regex.
# It specifies exactly how many times the preceding regex must occur for a match to succeed.
print(re.search('x-{3}x', 'x--x'))
# Two dashes: None
re.search('x-{3}x', 'x---x')
# Three dashes: <re.Match object; span=(0, 5), match='x---x'>
print(re.search('x-{3}x', 'x----x'))
# Four dashes: None
# The match fails when there are fewer or more than three dashes between the 'x' characters.

# {m,n} matches any number of repetitions of the preceding regex from m to n, inclusive.
# The match succeeds when there are two, three, or four dashes between the 'x'
# characters but fails otherwise.
for i in range(1, 6):
    s = f"x{'-' * i}x"
    print(f'{i}  {s:10}', re.search('x-{2,4}x', s))
# 1  x-x        None
# 2  x--x       <re.Match object; span=(0, 4), match='x--x'>
# 3  x---x      <re.Match object; span=(0, 5), match='x---x'>
# 4  x----x     <re.Match object; span=(0, 6), match='x----x'>
# 5  x-----x    None

# Omitting m implies a lower bound of 0, and omitting n implies an unlimited upper bound:
# <regex>{,n} is identical to <regex>{0,n}
# <regex>{m,} matches any number of repetitions greater than or equal to m
# <regex>{,}  is identical to <regex>{0,} and <regex>*

# If you omit all of m, n, and the comma, then the curly braces no longer function
# as metacharacters. {} matches just the literal string '{}'.
re.search('x{}y', 'x{}y')
# <re.Match object; span=(0, 4), match='x{}y'>

# To have any special meaning, a sequence with curly braces must fit one of these
# patterns: {m,n}, {m,}, {,n}, {,}. Otherwise, it matches literally.
re.search('x{foo}y', 'x{foo}y')
# <re.Match object; span=(0, 7), match='x{foo}y'>
re.search('x{a:b}y', 'x{a:b}y')
# <re.Match object; span=(0, 7), match='x{a:b}y'>
re.search('x{1,3,5}y', 'x{1,3,5}y')
# <re.Match object; span=(0, 9), match='x{1,3,5}y'>
re.search('x{foo,bar}y', 'x{foo,bar}y')
# <re.Match object; span=(0, 11), match='x{foo,bar}y'>

# {m,n}? is the non-greedy (lazy) version of {m,n}.
# {m,n} will match as many characters as possible, and {m,n}? will match as few as possible.
re.search('a{3,5}', 'aaaaaaaa')
# <re.Match object; span=(0, 5), match='aaaaa'>
re.search('a{3,5}?', 'aaaaaaaa')
# <re.Match object; span=(0, 3), match='aaa'>
# a{3,5} produces the longest possible match, so it matches five 'a' characters.
# a{3,5}? produces the shortest match, so it matches three.
```

## Grouping Constructs and Backreferences

```python
# Parentheses ( ) do two jobs:
# 1. Grouping: treat several characters as one unit (so a quantifier applies to all of it)
# 2. Capturing: remember what matched inside, so you can pull it out later

# On its own, a group matches the same thing as without parentheses
re.search('(bar)', 'foo bar baz')
# match='bar'
re.search('bar', 'foo bar baz')
# match='bar'
```

## Treating a Group as a Unit 

```python
# A quantifier metacharacter that follows a group operates on the entire
# subexpression specified in the group as a single unit.
# For instance, the following example matches one or more occurrences of the string 'bar'.
re.search('(bar)+', 'foo bar baz')
# <re.Match object; span=(4, 7), match='bar'>
re.search('(bar)+', 'foo barbar baz')
# <re.Match object; span=(4, 10), match='barbar'>
re.search('(bar)+', 'foo barbarbarbar baz')
# <re.Match object; span=(4, 16), match='barbarbarbar'>

# bar+   : The + metacharacter applies only to the character 'r'.
#          'ba' followed by one or more occurrences of 'r'
# (bar)+ : The + metacharacter applies to the entire string 'bar'.
#          One or more occurrences of 'bar'

# The regex (ba[rz]){2,4}(qux)? matches 2 to 4 occurrences of either 'bar' or
# 'baz', optionally followed by 'qux'.
re.search('(ba[rz]){2,4}(qux)?', 'bazbarbazqux')
# <re.Match object; span=(0, 12), match='bazbarbazqux'>
re.search('(ba[rz]){2,4}(qux)?', 'barbar')
# <re.Match object; span=(0, 6), match='barbar'>

# The following example shows that you can nest grouping parentheses.
re.search(r'(foo(bar)?)+(\d\d\d)?', 'foofoobar')
# <re.Match object; span=(0, 9), match='foofoobar'>
re.search(r'(foo(bar)?)+(\d\d\d)?', 'foofoobar123')
# <re.Match object; span=(0, 12), match='foofoobar123'>
re.search(r'(foo(bar)?)+(\d\d\d)?', 'foofoo123')
# <re.Match object; span=(0, 9), match='foofoo123'>
# String it all together and you get: at least one occurrence of 'foo' optionally
# followed by 'bar', all optionally followed by three decimal digit characters.
```

## Treating a Group as a Unit

```python
# A quantifier metacharacter that follows a group operates on the entire
# subexpression specified in the group as a single unit.
# For instance, the following example matches one or more occurrences of the string 'bar'.
re.search('(bar)+', 'foo bar baz')
# <re.Match object; span=(4, 7), match='bar'>
re.search('(bar)+', 'foo barbar baz')
# <re.Match object; span=(4, 10), match='barbar'>
re.search('(bar)+', 'foo barbarbarbar baz')
# <re.Match object; span=(4, 16), match='barbarbarbar'>

# bar+   : The + metacharacter applies only to the character 'r'.
#          'ba' followed by one or more occurrences of 'r'
# (bar)+ : The + metacharacter applies to the entire string 'bar'.
#          One or more occurrences of 'bar'

# The regex (ba[rz]){2,4}(qux)? matches 2 to 4 occurrences of either 'bar' or
# 'baz', optionally followed by 'qux'.
re.search('(ba[rz]){2,4}(qux)?', 'bazbarbazqux')
# <re.Match object; span=(0, 12), match='bazbarbazqux'>
re.search('(ba[rz]){2,4}(qux)?', 'barbar')
# <re.Match object; span=(0, 6), match='barbar'>

# The following example shows that you can nest grouping parentheses.
re.search(r'(foo(bar)?)+(\d\d\d)?', 'foofoobar')
# <re.Match object; span=(0, 9), match='foofoobar'>
re.search(r'(foo(bar)?)+(\d\d\d)?', 'foofoobar123')
# <re.Match object; span=(0, 12), match='foofoobar123'>
re.search(r'(foo(bar)?)+(\d\d\d)?', 'foofoo123')
# <re.Match object; span=(0, 9), match='foofoo123'>
# String it all together and you get: at least one occurrence of 'foo' optionally
# followed by 'bar', all optionally followed by three decimal digit characters.
```

## Capturing Groups

```python
# Most (but not quite all) grouping constructs also capture the part of the search
# string that matches the group. There are two methods defined for a match object
# that provide access to captured groups: .groups() and .group().

m = re.search(r'(\w+),(\w+),(\w+)', 'foo,quux,baz')
# Each of the three (\w+) expressions matches a sequence of word characters.

# m.groups() returns a tuple containing all the captured groups from a regex match.
m.groups()
# ('foo', 'quux', 'baz')
# The tuple contains the tokens but not the commas that appeared in the search string.

# m.group(<n>) returns a string containing the <n>th captured match.
# Note that the arguments are one-based, not zero-based.
m.group(1)
# 'foo'
m.group(2)
# 'quux'
m.group(3)
# 'baz'

# m.group(0) returns the entire match, and m.group() does the same.
m.group(0)
# 'foo,quux,baz'
m.group()
# 'foo,quux,baz'

# m.group(<n1>, <n2>, ...) returns a tuple containing the specified captured matches.
m.group(2, 3)
# ('quux', 'baz')
m.group(3, 2, 1)
# ('baz', 'quux', 'foo')
# This is just convenient shorthand. You could create the tuple of matches yourself instead.
(m.group(3), m.group(2), m.group(1))
# ('baz', 'quux', 'foo')
```

## Backreferences

```python
# You can match a previously captured group later within the same regex using a
# special metacharacter sequence called a backreference.
# \<n> matches the contents of the <n>th captured group.

# Here's a regex that matches a word, followed by a comma, followed by the same word again.
regex = r'(\w+),\1'

m = re.search(regex, 'foo,foo')
# <re.Match object; span=(0, 7), match='foo,foo'>
m.group(1)
# 'foo'

m = re.search(regex, 'qux,qux')
# <re.Match object; span=(0, 7), match='qux,qux'>
m.group(1)
# 'qux'

m = re.search(regex, 'foo,qux')
print(m)
# None
# What comes before the comma isn't the same as what comes after it, so the \1
# backreference doesn't match.

# Note: Any time you use a regex in Python with a numbered backreference, it's a
# good idea to specify it as a raw string. Otherwise, the interpreter may confuse
# the backreference with an octal value.
print(re.search('([a-z])#\1', 'd#d'))
# None
re.search(r'([a-z])#\1', 'd#d')
# <re.Match object; span=(0, 3), match='d#d'>
# Only the first ninety-nine captured groups are accessible by backreference.
```

## Other Grouping Constructs

```python
# (?P<name><regex>) creates a named captured group.
# The difference is that you reference the matched group by its given symbolic
# name instead of by its number.
m = re.search(r'(?P<w1>\w+),(?P<w2>\w+),(?P<w3>\w+)', 'foo,quux,baz')
m.groups()
# ('foo', 'quux', 'baz')
m.group('w1')
# 'foo'
m.group('w3')
# 'baz'
m.group('w1', 'w2', 'w3')
# ('foo', 'quux', 'baz')

# You can still access groups with symbolic names by number if you wish.
m.group(1)
# 'foo'
m.group(1, 2, 3)
# ('foo', 'quux', 'baz')

# (?P=<name>) matches the contents of a previously captured named group.
# It's a backreference, similar to \<n>, except that it refers to a named group
# rather than a numbered group.
m = re.search(r'(?P<word>\w+),(?P=word)', 'foo,foo')
# <re.Match object; span=(0, 7), match='foo,foo'>
m.group('word')
# 'foo'

# Note: The angle brackets (< and >) are required around name when creating a
# named group but not when referring to it later.
m = re.match(r'(?P<num>\d+)\.(?P=num)', '135.135')
# <re.Match object; span=(0, 7), match='135.135'>
m.group('num')
# '135'

# (?:<regex>) creates a non-capturing group.
# It matches the specified <regex>, but doesn't capture the match for later retrieval.
m = re.search(r'(\w+),(?:\w+),(\w+)', 'foo,quux,baz')
m.groups()
# ('foo', 'baz')
m.group(1)
# 'foo'
m.group(2)
# 'baz'
# The middle word 'quux' sits inside non-capturing parentheses, so it's missing
# from the tuple of captured groups.
# You may need the grouping feature, but you don't need to do anything with the
# value later, so you don't really need to capture it.

# (?(<n>)<yes-regex>|<no-regex>) specifies a conditional match.
# It matches against <yes-regex> if a group numbered <n> exists.
# Otherwise, it matches against <no-regex>.
regex = r'^(###)?foo(?(1)bar|baz)'
# ^(###)? indicates that the search string optionally begins with '###'.
# foo literally matches the string 'foo'.
# (?(1)bar|baz) matches against 'bar' if group 1 exists and 'baz' if it doesn't.

re.search(regex, '###foobar')
# <re.Match object; span=(0, 9), match='###foobar'>
print(re.search(regex, '###foobaz'))
# None
print(re.search(regex, 'foobar'))
# None
re.search(regex, 'foobaz')
# <re.Match object; span=(0, 6), match='foobaz'>

# Here's another conditional match using a named group instead of a numbered group.
# This regex matches the string 'foo', preceded by a single non-word character and
# followed by the same non-word character, or the string 'foo' by itself.
regex = r'^(?P<ch>\W)?foo(?(ch)(?P=ch)|)$'

re.search(regex, 'foo')
# <re.Match object; span=(0, 3), match='foo'>
re.search(regex, '#foo#')
# <re.Match object; span=(0, 5), match='#foo#'>
re.search(regex, '@foo@')
# <re.Match object; span=(0, 5), match='@foo@'>

print(re.search(regex, '#foo'))
# None
print(re.search(regex, 'foo@'))
# None
print(re.search(regex, '#foo@'))
# None
print(re.search(regex, '@foo#'))
# None
# Conditional regexes in Python are pretty esoteric and challenging to work through.
```

## Lookahead and Lookbehind Assertions 

```python
# Lookahead and lookbehind assertions determine the success or failure of a regex
# match based on what is just behind (to the left) or ahead (to the right) of the
# parser's current position in the search string.
# Like anchors, they are zero-width assertions, so they don't consume any of the
# search string. They also don't capture what they match.

# (?=<lookahead_regex>) creates a positive lookahead assertion.
# It asserts that what follows the regex parser's current position must match <lookahead_regex>.
re.search('foo(?=[a-z])', 'foobar')
# <re.Match object; span=(0, 3), match='foo'>
# The lookahead assertion (?=[a-z]) specifies that what follows 'foo' must be a
# lowercase alphabetic character. In this case, it's the character 'b'.

print(re.search('foo(?=[a-z])', 'foo123'))
# None
# The next character after 'foo' is '1', so there isn't a match.

# What's unique about a lookahead is that the portion of the search string that
# matches <lookahead_regex> isn't consumed, and it isn't part of the returned match object.
# Compare that to a similar example that uses grouping parentheses without a lookahead:
re.search('foo([a-z])', 'foobar')
# <re.Match object; span=(0, 4), match='foob'>
# This time, the regex consumes the 'b', and it becomes a part of the eventual match.

m = re.search('foo(?=[a-z])(?P<ch>.)', 'foobar')
m.group('ch')
# 'b'
# The lookahead matches 'b', but the parser doesn't advance past the 'b'.
# Then (?P<ch>.) matches the next single character available, which is 'b'.

m = re.search('foo([a-z])(?P<ch>.)', 'foobar')
m.group('ch')
# 'a'
# ([a-z]) matches and consumes 'b', so (?P<ch>.) matches the next character, which is now 'a'.

# (?!<lookahead_regex>) creates a negative lookahead assertion.
# It asserts that what follows the regex parser's current position must not match <lookahead_regex>.
re.search('foo(?=[a-z])', 'foobar')
# <re.Match object; span=(0, 3), match='foo'>
print(re.search('foo(?![a-z])', 'foobar'))
# None

print(re.search('foo(?=[a-z])', 'foo123'))
# None
re.search('foo(?![a-z])', 'foo123')
# <re.Match object; span=(0, 3), match='foo'>
# This is the opposite of what happened with the corresponding positive lookahead assertions.

# (?<=<lookbehind_regex>) creates a positive lookbehind assertion.
# It asserts that what precedes the regex parser's current position must match <lookbehind_regex>.
re.search('(?<=foo)bar', 'foobar')
# <re.Match object; span=(3, 6), match='bar'>
print(re.search('(?<=qux)bar', 'foobar'))
# None
# The lookbehind requires that 'qux' precede 'bar'.

# The <lookbehind_regex> in a lookbehind assertion must specify a match of fixed length.
# This isn't allowed because the length of the string matched by a+ is indeterminate:
# re.search('(?<=a+)def', 'aaadef')
# error: look-behind requires fixed-width pattern

# This, however, is okay:
re.search('(?<=a{3})def', 'aaadef')
# <re.Match object; span=(3, 6), match='def'>
# Anything that matches a{3} will have a fixed length of three.

# (?<!<lookbehind_regex>) creates a negative lookbehind assertion.
# It asserts that what precedes the regex parser's current position must not match <lookbehind_regex>.
print(re.search('(?<!foo)bar', 'foobar'))
# None
re.search('(?<!qux)bar', 'foobar')
# <re.Match object; span=(3, 6), match='bar'>
```

## Miscellaneous Metacharacters

```python
# (?#...) specifies a comment.
# The regex parser ignores anything contained in the sequence (?#...).
re.search('bar(?#This is a comment) *baz', 'foo bar baz qux')
# <re.Match object; span=(4, 11), match='bar baz'>
# This allows you to specify documentation inside a regex in Python.

# Vertical bar, or pipe (|), specifies a set of alternatives on which to match.
# An expression of the form <regex1>|<regex2>|...|<regexn> matches at most one of
# the specified expressions.
re.search('foo|bar|baz', 'bar')
# <re.Match object; span=(0, 3), match='bar'>
re.search('foo|bar|baz', 'baz')
# <re.Match object; span=(0, 3), match='baz'>
print(re.search('foo|bar|baz', 'quux'))
# None

# Alternation is non-greedy. The regex parser looks at the expressions separated
# by | in left-to-right order and returns the first match that it finds.
re.search('foo', 'foograult')
# <re.Match object; span=(0, 3), match='foo'>
re.search('grault', 'foograult')
# <re.Match object; span=(3, 9), match='grault'>
re.search('foo|grault', 'foograult')
# <re.Match object; span=(0, 3), match='foo'>
# The match returned is 'foo' because that appears first when scanning from left
# to right, even though 'grault' would be a longer match.

# You can combine alternation, grouping, and any other metacharacters.
# (foo|bar|baz)+ means a sequence of one or more of the strings 'foo', 'bar', or 'baz'.
re.search('(foo|bar|baz)+', 'foofoofoo')
# <re.Match object; span=(0, 9), match='foofoofoo'>
re.search('(foo|bar|baz)+', 'bazbazbazbaz')
# <re.Match object; span=(0, 12), match='bazbazbazbaz'>
re.search('(foo|bar|baz)+', 'barbazfoo')
# <re.Match object; span=(0, 9), match='barbazfoo'>

# ([0-9]+|[a-f]+) means a sequence of one or more decimal digit characters or a
# sequence of one or more of the characters 'a-f'.
re.search('([0-9]+|[a-f]+)', '456')
# <re.Match object; span=(0, 3), match='456'>
re.search('([0-9]+|[a-f]+)', 'ffda')
# <re.Match object; span=(0, 4), match='ffda'>
```

## Modified Regular Expression Matching With Flags

```python
# Most of the functions in the re module take an optional <flags> argument.
# re.search(<regex>, <string>, <flags>) scans a string for a regex match,
# applying the specified modifier <flags>.
# Flags modify regex parsing behavior, allowing you to refine your pattern matching even further.
```

## Supported Regular Expression Flags

```python
# All flags except re.DEBUG have a short, single-letter name and also a longer, full-word name:
# re.I  re.IGNORECASE  Makes matching of alphabetic characters case-insensitive
# re.M  re.MULTILINE   Causes start-of-string and end-of-string anchors to match embedded newlines
# re.S  re.DOTALL      Causes the dot metacharacter to match a newline
# re.X  re.VERBOSE     Allows inclusion of whitespace and comments within a regular expression
#       re.DEBUG       Causes the regex parser to display debugging information to the console
# re.A  re.ASCII       Specifies ASCII encoding for character classification
# re.U  re.UNICODE     Specifies Unicode encoding for character classification
# re.L  re.LOCALE      Specifies encoding for character classification based on the current locale

# re.I / re.IGNORECASE makes matching case insensitive.
re.search('a+', 'aaaAAA')
# <re.Match object; span=(0, 3), match='aaa'>
re.search('A+', 'aaaAAA')
# <re.Match object; span=(3, 6), match='AAA'>
re.search('a+', 'aaaAAA', re.I)
# <re.Match object; span=(0, 6), match='aaaAAA'>
re.search('A+', 'aaaAAA', re.IGNORECASE)
# <re.Match object; span=(0, 6), match='aaaAAA'>
# In the subsequent searches, the parser ignores case, so both a+ and A+ match the entire string.

# IGNORECASE affects alphabetic matching involving character classes as well.
re.search('[a-z]+', 'aBcDeF')
# <re.Match object; span=(0, 1), match='a'>
re.search('[a-z]+', 'aBcDeF', re.I)
# <re.Match object; span=(0, 6), match='aBcDeF'>

# re.M / re.MULTILINE causes start-of-string and end-of-string anchors to match at embedded newlines.
# By default, the ^ and $ anchors match only at the beginning and end of the search string.
s = 'foo\nbar\nbaz'

re.search('^foo', s)
# <re.Match object; span=(0, 3), match='foo'>
print(re.search('^bar', s))
# None
print(re.search('^baz', s))
# None

print(re.search('foo$', s))
# None
print(re.search('bar$', s))
# None
re.search('baz$', s)
# <re.Match object; span=(8, 11), match='baz'>

# If the MULTILINE flag is set, the ^ and $ anchor metacharacters match internal lines as well:
# ^ matches at the beginning of the string or at the beginning of any line within the string.
# $ matches at the end of the string or at the end of any line within the string.
re.search('^foo', s, re.MULTILINE)
# <re.Match object; span=(0, 3), match='foo'>
re.search('^bar', s, re.MULTILINE)
# <re.Match object; span=(4, 7), match='bar'>
re.search('^baz', s, re.MULTILINE)
# <re.Match object; span=(8, 11), match='baz'>

re.search('foo$', s, re.M)
# <re.Match object; span=(0, 3), match='foo'>
re.search('bar$', s, re.M)
# <re.Match object; span=(4, 7), match='bar'>
re.search('baz$', s, re.M)
# <re.Match object; span=(8, 11), match='baz'>

# Note: The MULTILINE flag only modifies the ^ and $ anchors in this way.
# It doesn't have any effect on the \A and \Z anchors.
print(re.search(r'\Abar', s, re.MULTILINE))
# None
print(re.search(r'bar\Z', s, re.MULTILINE))
# None

# re.S / re.DOTALL causes the dot (.) metacharacter to match a newline.
# By default, the dot metacharacter matches any character except the newline
# character. The DOTALL flag lifts this restriction.
print(re.search('foo.bar', 'foo\nbar'))
# None
re.search('foo.bar', 'foo\nbar', re.DOTALL)
# <re.Match object; span=(0, 7), match='foo\nbar'>
re.search('foo.bar', 'foo\nbar', re.S)
# <re.Match object; span=(0, 7), match='foo\nbar'>
# Note that the short name of the DOTALL flag is re.S, not re.D as you might expect.

# re.X / re.VERBOSE allows inclusion of whitespace and comments within a regex.
# The regex parser ignores all whitespace unless it's within a character class or
# escaped with a backslash. It also ignores a # character and all characters to the right of it.
# Suppose you want to parse phone numbers with an optional three-digit area code in
# parentheses, optional whitespace, a three-digit prefix, a separator ('-' or '.'),
# and a four-digit line number.
regex = r'^(\(\d{3}\))?\s*\d{3}[-.]\d{4}$'

re.search(regex, '414.9229')
# <re.Match object; span=(0, 8), match='414.9229'>
re.search(regex, '414-9229')
# <re.Match object; span=(0, 8), match='414-9229'>
re.search(regex, '(712)414-9229')
# <re.Match object; span=(0, 13), match='(712)414-9229'>
re.search(regex, '(712) 414-9229')
# <re.Match object; span=(0, 14), match='(712) 414-9229'>

# But r'^(\(\d{3}\))?\s*\d{3}[-.]\d{4}$' is an eyeful, isn't it?
# Using the VERBOSE flag, you can write the same regex in Python like this instead:
regex = r'''^               # Start of string
            (\(\d{3}\))?    # Optional area code
            \s*             # Optional whitespace
            \d{3}           # Three-digit prefix
            [-.]            # Separator character
            \d{4}           # Four-digit line number
            $               # Anchor at end of string
            '''

re.search(regex, '414.9229', re.VERBOSE)
# <re.Match object; span=(0, 8), match='414.9229'>
re.search(regex, '414-9229', re.VERBOSE)
# <re.Match object; span=(0, 8), match='414-9229'>
re.search(regex, '(712)414-9229', re.X)
# <re.Match object; span=(0, 13), match='(712)414-9229'>
re.search(regex, '(712) 414-9229', re.X)
# <re.Match object; span=(0, 14), match='(712) 414-9229'>

# When using the VERBOSE flag, be mindful of whitespace that you do intend to be significant.
re.search('foo bar', 'foo bar')
# <re.Match object; span=(0, 7), match='foo bar'>
print(re.search('foo bar', 'foo bar', re.VERBOSE))
# None
# The VERBOSE flag causes the parser to ignore the space character.

# To make this match as expected, escape the space character with a backslash or
# include it in a character class.
re.search(r'foo\ bar', 'foo bar', re.VERBOSE)
# <re.Match object; span=(0, 7), match='foo bar'>
re.search('foo[ ]bar', 'foo bar', re.VERBOSE)
# <re.Match object; span=(0, 7), match='foo bar'>
# The VERBOSE flag has a non-intuitive short name: re.X, not re.V.

# re.DEBUG displays debugging information.
re.search('foo.bar', 'fooxbar', re.DEBUG)
# LITERAL 102
# LITERAL 111
# LITERAL 111
# ANY None
# LITERAL 98
# LITERAL 97
# LITERAL 114
# <re.Match object; span=(0, 7), match='fooxbar'>
# When the parser displays LITERAL nnn in the debugging output, it's showing the
# ASCII code of a literal character in the regex.

# Deep Dive: MAX_REPEAT 2 4 confirms that the regex parser recognizes the
# metacharacter sequence {2,4} and interprets it as a range quantifier.
re.search('x[123]{2,4}y', 'x222y', re.DEBUG)
# There's no MAX_REPEAT token here, so the parser treats {foo} literally.
re.search('x[123]{foo}y', 'x222y', re.DEBUG)

# The re module doesn't define a single-letter version of the DEBUG flag.
# You could define your own if you wanted to, but readers might misconstrue it
# as an abbreviation for the DOTALL flag.
re.D = re.DEBUG
re.search('foo', 'foo', re.D)

# re.A / re.ASCII, re.U / re.UNICODE, re.L / re.LOCALE specify the character
# encoding used for parsing of special regex character classes (\w, \W, \b, \B, \d, \D, \s, \S).

# For the regex parser to properly account for the Devanagari script, the digit
# metacharacter sequence \d must match each of these characters as well.
s = '\u0967\u096a\u096c'
# '१४६'
re.search(r'\d+', s)
# <re.Match object; span=(0, 3), match='१४६'>

s = 'sch\u00f6n'
# 'schön'
re.search(r'\w+', s, re.ASCII)
# <re.Match object; span=(0, 3), match='sch'>
# When you restrict the encoding to ASCII, the regex parser recognizes only the
# first three characters as word characters. The match stops at 'ö'.
re.search(r'\w+', s, re.UNICODE)
# <re.Match object; span=(0, 5), match='schön'>
re.search(r'\w+', s)
# <re.Match object; span=(0, 5), match='schön'>
# In general, the best strategy is to use the default Unicode encoding.
```

## Combining flags Arguments in a Function Call

```python
# Flag values are defined so that you can combine them using the bitwise OR (|)
# operator. This allows you to specify several flags in a single function call.
re.search('^bar', 'FOO\nBAR\nBAZ', re.I|re.M)
# <re.Match object; span=(4, 7), match='BAR'>
# This re.search() call uses bitwise OR to specify both the IGNORECASE and
# MULTILINE flags at once.
```

## Setting and Clearing Flags Within a Regular Expression

```python
# (?<flags>) sets flag value(s) for the duration of a regex.
# The value of <flags> is one or more letters from the set a, i, L, m, s, u, and x:
# a = re.ASCII, i = re.IGNORECASE, L = re.LOCALE, m = re.MULTILINE,
# s = re.DOTALL, u = re.UNICODE, x = re.VERBOSE

# The following examples are equivalent ways of setting the IGNORECASE and MULTILINE flags.
re.search('^bar', 'FOO\nBAR\nBAZ\n', re.I|re.M)
# <re.Match object; span=(4, 7), match='BAR'>
re.search('(?im)^bar', 'FOO\nBAR\nBAZ\n')
# <re.Match object; span=(4, 7), match='BAR'>

# A (?<flags>) metacharacter sequence sets the given flag(s) for the entire regex.
# As of Python 3.7, it's deprecated to specify (?<flags>) anywhere in a regex
# other than at the beginning. In Python 3.11 and newer, it has to go at the start
# or you get: PatternError: global flags not at the start of the expression
re.search('(?s)foo.bar.baz', 'foo\nbar\nbaz')
# <re.Match object; span=(0, 11), match='foo\nbar\nbaz'>
# Both dot metacharacters match newlines because the DOTALL flag is in effect.

# (?<set_flags>-<remove_flags>:<regex>) sets or removes flag value(s) for the
# duration of a group.
re.search('(?i:foo)bar', 'FOObar')
# <re.Match object; span=(0, 6), match='FOObar'>
# This produces a match because (?i:foo) dictates that the match against 'FOO' is case insensitive.

print(re.search('(?i:foo)bar', 'FOOBAR'))
# None
# Once outside the group, IGNORECASE is no longer in effect, so the match against
# 'BAR' is case sensitive and fails.

# Here's an example that demonstrates turning a flag off for a group.
print(re.search('(?-i:foo)bar', 'FOOBAR', re.IGNORECASE))
# None
# (?-i:foo) turns off IGNORECASE for the duration of that group, so the match against 'FOO' fails.

# As of Python 3.7, you can specify u, a, or L as <set_flags> to override the
# default encoding for the specified group.
s = 'sch\u00f6n'
re.search(r'(?a:\w+)', s)
# <re.Match object; span=(0, 3), match='sch'>
re.search(r'(?u:\w+)', s)
# <re.Match object; span=(0, 5), match='schön'>

# You can only set encoding this way, though. You can't remove it:
# re.search(r'(?-a:\w+)', s)
# re.error: bad inline flags: cannot turn off flags 'a', 'u' and 'L'
# u, a, and L are mutually exclusive. Only one of them may appear per group.
```
